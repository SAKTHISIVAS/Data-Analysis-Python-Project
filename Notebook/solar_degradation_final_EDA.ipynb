### 5.2 Timestamp Validation and Conversion

The timestamp column identifies when each solar PV measurement was recorded. It is initially stored as a string, which limits time-based analysis.

In this step, we convert the timestamp column to a datetime data type and check whether any values fail to convert. The original DataFrame is preserved, and the converted values are stored in a separate working copy.

```python
# Step 6.4: Convert timestamps in a separate working copy

# Preserve the original dataset
prep_df = df.copy()

# Convert timestamp strings to datetime values
prep_df['timestamp'] = pd.to_datetime(
    prep_df['timestamp'],
    errors='coerce'
)

# Check the conversion results
print("Timestamp data type:", prep_df['timestamp'].dtype)
print("Invalid timestamps:", prep_df['timestamp'].isna().sum())

# Display the earliest and latest timestamps
print("Earliest timestamp:", prep_df['timestamp'].min())
print("Latest timestamp:", prep_df['timestamp'].max())

# Display a few converted timestamps
display(prep_df[['timestamp']].head())
```

### 5.3 Timestamp Interval and Continuity Checks

Solar PV measurements are expected to follow a regular sampling schedule. In this step, we examine the time differences between consecutive timestamps to identify the sampling interval and possible gaps.

This check helps assess temporal consistency and determine whether the dataset contains missing measurement periods. Gaps will be reported for investigation rather than automatically filled.

```python
# Step 6.5: Check timestamp intervals

# Sort records chronologically in the working copy
prep_df = prep_df.sort_values('timestamp').copy()

# Calculate the time difference between consecutive records
prep_df['time_difference'] = prep_df['timestamp'].diff()

# Display the most common time intervals
print("Most common timestamp intervals:")
display(prep_df['time_difference'].value_counts().head(10))

# Count intervals that differ from 15 minutes
expected_interval = pd.Timedelta(minutes=15)

interval_counts = {
    "Total records": len(prep_df),
    "Intervals examined": prep_df['time_difference'].notna().sum(),
    "Intervals not equal to 15 minutes": (
        (prep_df['time_difference'].notna()) &
        (prep_df['time_difference'] != expected_interval)
    ).sum()
}

display(pd.Series(interval_counts))
```

### 5.6 Date and Time Consistency Checks

The dataset contains a timestamp column and additional date-related variables. These fields should represent consistent information.

In this step, we compare the date, hour, month, and year columns with values derived from the converted timestamp. Any discrepancies will be reported for investigation. We will not overwrite the original values automatically.

```python
# Step 6.8: Compare date-related columns with the timestamp

# Derive expected values from the converted timestamp
expected_date = prep_df['timestamp'].dt.strftime('%Y-%m-%d')
expected_hour = prep_df['timestamp'].dt.hour
expected_month = prep_df['timestamp'].dt.month
expected_year = prep_df['timestamp'].dt.year

# Compare existing columns with timestamp-derived values
date_checks = pd.Series({
    'Date mismatches': (
        prep_df['date'].astype('string').str[:10]
        != expected_date
    ).fillna(False).sum(),

    'Hour mismatches': (
        prep_df['hour'] != expected_hour
    ).fillna(False).sum(),

    'Month mismatches': (
        prep_df['month'] != expected_month
    ).fillna(False).sum(),

    'Year mismatches': (
        prep_df['year'] != expected_year
    ).fillna(False).sum()
})

display(date_checks)
```

### 5.7 Investigation of Date and Time Inconsistencies

The consistency checks identified one potential mismatch in each of the hour, month, and year columns. In this step, we inspect records with missing or inconsistent date-related values to determine the cause.

Any corrections will be based on the validated timestamp and documented before being applied to the working dataset.

```python
# Step 6.9: Identify missing or inconsistent date-related values

# Derive expected date and time values from the timestamp
expected_hour = prep_df['timestamp'].dt.hour
expected_month = prep_df['timestamp'].dt.month
expected_year = prep_df['timestamp'].dt.year

# Identify rows with missing or mismatched values
issue_mask = (
    prep_df['date'].isna()
    | prep_df['hour'].isna()
    | prep_df['month'].isna()
    | prep_df['year'].isna()
    | (prep_df['hour'] != expected_hour)
    | (prep_df['month'] != expected_month)
    | (prep_df['year'] != expected_year)
)

# Display the affected records for investigation
display(
    prep_df.loc[
        issue_mask,
        ['timestamp', 'date', 'hour', 'month', 'year']
    ]
)
```

### 5.8 Correction of Missing Date and Time Attributes

The investigation identified one record with a valid timestamp but missing hour, month, and year values. Since these fields can be derived directly from the validated timestamp, they can be reconstructed without estimating or inventing measurements.

The corrections will be applied only to the working DataFrame. The original dataset will remain unchanged.

```python
# Step 6.10: Reconstruct missing date and time attributes

# Fill missing values using the validated timestamp
prep_df['date'] = prep_df['date'].fillna(
    prep_df['timestamp'].dt.strftime('%Y-%m-%d')
)

prep_df['hour'] = prep_df['hour'].fillna(
    prep_df['timestamp'].dt.hour
)

prep_df['month'] = prep_df['month'].fillna(
    prep_df['timestamp'].dt.month
)

prep_df['year'] = prep_df['year'].fillna(
    prep_df['timestamp'].dt.year
)

# Verify that the missing fields have been corrected
print("Missing date values:", prep_df['date'].isna().sum())
print("Missing hour values:", prep_df['hour'].isna().sum())
print("Missing month values:", prep_df['month'].isna().sum())
print("Missing year values:", prep_df['year'].isna().sum())

# Inspect the corrected record
display(
    prep_df.loc[
        prep_df['timestamp'].dt.strftime('%Y-%m-%d %H:%M:%S')
        == '2024-03-08 01:00:00',
        ['timestamp', 'date', 'hour', 'month', 'year']
    ]
)
```

### 5.9 Investigation of Zero Power Readings

Zero power readings are not necessarily errors in solar photovoltaic data. They can occur naturally during nighttime or low-irradiance conditions.

In this step, we count zero readings in total inverter power and individual inverter channels. We also compare these readings with solar irradiance to identify observations that may require further investigation.

No zero values will be removed or replaced automatically.

```python
# Step 6.11: Count zero power readings

# Identify individual inverter power columns
inverter_cols = [
    col for col in prep_df.columns
    if col.startswith('inv_')
]

# Include total inverter power
power_cols = ['total_inverter_ac_power'] + inverter_cols

# Count zero readings in each power column
zero_counts = (
    prep_df[power_cols]
    .eq(0)
    .sum()
    .sort_values(ascending=False)
)

zero_summary = pd.DataFrame({
    'Zero Readings': zero_counts,
    'Zero Percentage (%)': (
        zero_counts / len(prep_df) * 100
    ).round(2)
})

display(zero_summary)
```

```python
# Step 6.11: Investigate zero power when irradiance is high enough

poa_col = 'poa_irradiance_o_149574'

# Select records with POA irradiance >= 200 W/m²
solar_active_mask = prep_df[poa_col] >= 200

print("Solar-active records:", solar_active_mask.sum())

# Count zero total-power readings within the selected records
zero_total_power_active = (
    prep_df.loc[solar_active_mask, 'total_inverter_ac_power']
    .eq(0)
    .sum()
)

print(
    "Zero total-power readings during solar-active conditions:",
    zero_total_power_active
)
```

```python
# Step 5.14: Identify consecutive missing-irradiance intervals

# Create a Boolean indicator for missing POA irradiance
missing_mask = prep_df[poa_col].isna()

# Identify the beginning of each consecutive group
group_start = missing_mask.ne(missing_mask.shift()).cumsum()

# Count the length of each consecutive missing group
missing_group_sizes = (
    prep_df.loc[missing_mask]
    .groupby(group_start[missing_mask])
    .size()
)

print("Number of consecutive missing-data groups:",
      len(missing_group_sizes))

print("\nLargest consecutive missing-data groups (in 15-minute records):")
display(
    missing_group_sizes
    .sort_values(ascending=False)
    .head(10)
    .rename('Consecutive Missing Records')
    .to_frame()
)

print("\nTotal missing records across all groups:",
      missing_group_sizes.sum())
```

```python
# Step 5.14: Summarize consecutive missing POA irradiance groups

missing_mask = prep_df[poa_col].isna()

# Create a new group whenever the missing/non-missing status changes
group_id = missing_mask.ne(missing_mask.shift()).cumsum()

# Count consecutive missing records in each group
missing_group_sizes = (
    prep_df.loc[missing_mask]
    .groupby(group_id[missing_mask])
    .size()
)

print("Number of missing-data groups:", len(missing_group_sizes))
print("Largest group size:", missing_group_sizes.max())
print("Total missing records:", missing_group_sizes.sum())

print("\nAll group sizes:")
display(missing_group_sizes.reset_index(drop=True).to_frame(
    name="Consecutive Missing Records"
))
```

### Step 10.3: POA Irradiance Correlation by Inverter

In this step, we calculate the Pearson correlation between POA irradiance and the power output of each inverter.

**Purpose:**
- Compare how closely each inverter's output tracks irradiance.
- Identify inverters with unusually weak linear associations.
- Generate evidence for further investigation of differences between inverters.

**Important:** A lower correlation does not automatically indicate a faulty inverter. Missing readings, zero-output periods, operating conditions, and other factors can influence correlation values.

```python
# Step 10.3: Calculate POA irradiance correlation for every inverter

inverter_irradiance_corr = (
    prep_df[inverter_cols + [poa_col]]
    .corr(method='pearson')[poa_col]
    .drop(labels=[poa_col])
    .sort_values()
)

correlation_summary = inverter_irradiance_corr.rename(
    'POA_Irradiance_Correlation'
).to_frame()

display(correlation_summary.round(3))

print("Lowest correlation:",
      correlation_summary['POA_Irradiance_Correlation'].min())
print("Highest correlation:",
      correlation_summary['POA_Irradiance_Correlation'].max())
```

The analysis compares the Pearson correlation between POA irradiance and the power output of all 24 inverters.

- **Other inverters:** Have correlations ranging approximately from 0.990 to 0.994.

**Conclusion:** INV-16 behaves differently from the rest of the inverter fleet in its relationship with irradiance. This observation is consistent with the earlier analysis showing lower output from INV-16 during high-irradiance periods.

**Limitation:** A lower correlation does not prove equipment failure or degradation. Further investigation would require operating-condition checks and a suitable expected-power model.

## 11. Anomaly Exploration

Anomaly exploration helps identify observations that differ from expected or typical behavior.

In this project, we focus on inverter output during periods of relatively high POA irradiance. These periods are useful for examining whether individual inverters produce substantially less power than the rest of the fleet.

**Important:** A statistical anomaly is a candidate for investigation, not proof of equipment failure or solar panel degradation.

### Step 11.1: Compare Inverter Output During High Irradiance

We use the existing high-irradiance threshold of POA irradiance greater than or equal to 200. For each inverter, we calculate its mean power during these observations and compare it with the fleet average.

This helps identify unusually low average output under similar irradiance conditions.

```python
# Step 11.1: Compare inverter output during high irradiance

high_irradiance_data = prep_df.loc[solar_active_mask]

inverter_high_irradiance_mean = (
    high_irradiance_data[inverter_cols].mean()
)

fleet_mean = inverter_high_irradiance_mean.mean()

high_irradiance_summary = pd.DataFrame({
    'Average_Inverter_Power': inverter_high_irradiance_mean,
    'Fleet_Average_Power': fleet_mean
})

high_irradiance_summary['Difference_From_Fleet_Percent'] = (
    (
        high_irradiance_summary['Average_Inverter_Power']
        - fleet_mean
    ) / fleet_mean * 100
)

high_irradiance_summary = high_irradiance_summary.sort_values(
    'Difference_From_Fleet_Percent'
)

display(high_irradiance_summary.round(2))
```

### Interpretation: Inverter Output During High Irradiance

The analysis compares the average power of all 24 inverters when POA irradiance is at least 200.

**Key observations:**

- **Fleet average:** The mean across the 24 inverter-level averages is 17.73.
- **INV-16:** Average power is 8.46, approximately 52.26% below the fleet average.
- **INV-10:** Average power is 14.43, approximately 18.60% below the fleet average.
- **INV-08:** Average power is 16.37, approximately 7.67% below the fleet average.
- **Remaining inverters:** Their average power ranges from approximately 17.96 to 18.68, placing them above the fleet average.

**Conclusion:** INV-16 shows the largest difference below the fleet average, followed by INV-10 and INV-08. These inverters are candidates for further performance investigation.

**Limitation:** This comparison uses raw power measurements and does not account for all possible differences in inverter capacity, configuration, or operating conditions. A below-average result does not independently establish a fault or degradation.

### Step 11.2: Visualizing Inverter Performance Differences

This chart displays each inverter's percentage difference from the fleet average during high-irradiance periods.

Values below zero indicate average power below the fleet average, while values above zero indicate average power above it. The chart is intended to identify candidates for further investigation, not to diagnose equipment faults.

```python
# Step 11.2: Plot percentage difference from fleet average

difference_values = high_irradiance_summary[
    'Difference_From_Fleet_Percent'
].sort_values()

plt.figure(figsize=(12, 8))

difference_values.plot(kind='barh')

plt.axvline(0, linestyle='--', linewidth=1)

plt.title('Inverter Power Difference from Fleet Average')
plt.xlabel('Difference from Fleet Average (%)')
plt.ylabel('Inverter')
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()
```

**Key Findings — Inverter Anomaly Exploration**

The inverter comparison under high-irradiance conditions identified differences in power output across the solar system. INV-16 showed the largest deviation, with average power approximately 52.26% below the fleet average. INV-10 was approximately 18.60% below the fleet average, while INV-08 was approximately 7.67% below.

These inverters are candidates for further investigation. However, the observed differences alone do not establish equipment failure or long-term degradation. Additional operational data and longer-term analysis would be needed to investigate the underlying causes.
